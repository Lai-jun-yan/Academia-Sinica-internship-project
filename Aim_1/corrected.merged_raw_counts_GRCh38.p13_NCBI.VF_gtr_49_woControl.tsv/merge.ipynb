```python
import pandas as pd

file1 = r"C:\Users\USER\Desktop\資訊所實習\計畫\Gene list\corrected.merged_raw_counts_GRCh38.p13_NCBI.VF_gtr_49_woControl.tsv\corrected.merged_raw_counts_GRCh38.p13_NCBI.VF_gtr_49_woControl.tsv"

expr_df = pd.read_csv(
    file1,
    sep="\t"
)

print(expr_df.head())
```

```python
file2 = r"C:\Users\USER\Desktop\資訊所實習\計畫\Gene list\corrected.merged_raw_counts_GRCh38.p13_NCBI.VF_gtr_49_woControl.tsv\Human.GRCh38.p13.annot.tsv"

gene_df = pd.read_csv(
    file2,
    sep="\t"
)

print(gene_df.head())
```

```python
expr_df["GeneID"] = expr_df["GeneID"].astype(str)
gene_df["GeneID"] = gene_df["GeneID"].astype(str)
```

```python
annotation = gene_df[["GeneID", "Symbol"]]
```

```python
merged = expr_df.merge(
    annotation,
    on="GeneID",
    how="left"
)
```

```python
print(merged.head(10))
```

```python
merged = merged.dropna(subset=["Symbol"])

merged["Symbol"].isna().sum()
```

```python
merged["GeneID"].duplicated().sum()
```

```python
merged = merged.set_index("GeneID")

cols = ["Symbol"] + [
    c for c in merged.columns 
    if c != "Symbol"
]

merged = merged[cols]
```

```python
print(merged.head())
```

```python
print(merged[
    merged["Symbol"].duplicated(keep=False)
].sort_values("Symbol"))
```

```python
dup_symbol = merged.loc[
    merged["Symbol"].duplicated(keep=False),
    "Symbol"
].unique()

merged = merged[
    ~merged["Symbol"].isin(dup_symbol)
]
```

```python
print("shape:", merged.shape)
print(
    "duplicate Symbol:",
    merged["Symbol"].duplicated().sum()
)
```

```python
print(merged.shape)

print("duplicate GeneID:",
      merged.index.duplicated().sum())

print("duplicate Symbol:",
      merged["Symbol"].duplicated().sum())

print("missing Symbol:",
      merged["Symbol"].isna().sum())
```

```python
Alphaviridae_gene_df = pd.read_csv(
    r"C:\Users\USER\Desktop\資訊所實習\計畫\Gene list\immune\consensus_results.Alphaviridae.txt",
    sep="\t"
)

Coronaviridae_gene_df = pd.read_csv(
    r"C:\Users\USER\Desktop\資訊所實習\計畫\Gene list\immune\consensus_results.Coronaviridae.txt",
    sep="\t"
)

Flaviviridae_gene_df = pd.read_csv(
    r"C:\Users\USER\Desktop\資訊所實習\計畫\Gene list\immune\consensus_results.Flaviviridae.txt",
    sep="\t"
)

Orthoherpesviridae_gene_df = pd.read_csv(
    r"C:\Users\USER\Desktop\資訊所實習\計畫\Gene list\immune\consensus_results.Orthoherpesviridae.txt",
    sep="\t"
)

Orthomyxoviridae_gene_df = pd.read_csv(
    r"C:\Users\USER\Desktop\資訊所實習\計畫\Gene list\immune\consensus_results.Orthomyxoviridae.txt",
    sep="\t"
)

Picornaviridae_gene_df = pd.read_csv(
    r"C:\Users\USER\Desktop\資訊所實習\計畫\Gene list\immune\consensus_results.Picornaviridae.txt",
    sep="\t"
)

Reoviridae_gene_df = pd.read_csv(
    r"C:\Users\USER\Desktop\資訊所實習\計畫\Gene list\immune\consensus_results.Reoviridae.txt",
    sep="\t"
)

Retroviridae_gene_df = pd.read_csv(
    r"C:\Users\USER\Desktop\資訊所實習\計畫\Gene list\immune\consensus_results.Retroviridae.txt",
    sep="\t"
)
```

```python
# 挑 RF score 不為 0 的 gene
Alphaviridae_selected_genes = Alphaviridae_gene_df.loc[
    Alphaviridae_gene_df["RF_mean_abs_Score"] != 0,
    "Gene"
].tolist()

Coronaviridae_selected_genes = Coronaviridae_gene_df.loc[
    Coronaviridae_gene_df["RF_mean_abs_Score"] != 0,
    "Gene"
].tolist()

Flaviviridae_selected_genes = Flaviviridae_gene_df.loc[
    Flaviviridae_gene_df["RF_mean_abs_Score"] != 0,
    "Gene"
].tolist()

Orthoherpesviridae_selected_genes = Orthoherpesviridae_gene_df.loc[
    Orthoherpesviridae_gene_df["RF_mean_abs_Score"] != 0,
    "Gene"
].tolist()

Orthomyxoviridae_selected_genes = Orthomyxoviridae_gene_df.loc[
    Orthomyxoviridae_gene_df["RF_mean_abs_Score"] != 0,
    "Gene"
].tolist()

Picornaviridae_selected_genes = Picornaviridae_gene_df.loc[
    Picornaviridae_gene_df["RF_mean_abs_Score"] != 0,
    "Gene"
].tolist()

Reoviridae_selected_genes = Reoviridae_gene_df.loc[
    Reoviridae_gene_df["RF_mean_abs_Score"] != 0,
    "Gene"
].tolist()

Retroviridae_selected_genes = Retroviridae_gene_df.loc[
    Retroviridae_gene_df["RF_mean_abs_Score"] != 0,
    "Gene"
].tolist()

```

```python
# 取聯集
selected_genes = list(
    set(Alphaviridae_selected_genes)
    | set(Coronaviridae_selected_genes)
    | set(Flaviviridae_selected_genes)
    | set(Orthoherpesviridae_selected_genes)
    | set(Orthomyxoviridae_selected_genes)
    | set(Picornaviridae_selected_genes)
    | set(Reoviridae_selected_genes)
    | set(Retroviridae_selected_genes)
)
```

```python
print("非0 gene 數量:", len(selected_genes))
print(selected_genes[:10])
```

```python
wgcna_selected = wgcna[
    wgcna.columns.intersection(selected_genes)
]
```

```python
print(wgcna_selected.shape)
```

```python
missing_genes = set(selected_genes) - set(wgcna.columns)

print("gene list 數量:", len(selected_genes))
print("WGCNA找到:", wgcna_selected.shape[1])
print("找不到:", len(missing_genes))
```
